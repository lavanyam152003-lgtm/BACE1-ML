```python
from rdkit.Chem import Descriptors
from rdkit import Chem
import pandas as pd
def calculate_descriptors(smiles):
    mol=Chem.MolFromSmiles(smiles)
    return {
    "MolWt":Descriptors.MolWt(mol),
    "LogP":Descriptors.MolLogP(mol),
    "NumHDonors":Descriptors.NumHDonors(mol),
    "NumHAcceptors":Descriptors.NumHAcceptors(mol),
    "TPSA":Descriptors.TPSA(mol),
    "RotatableBonds":Descriptors.NumRotatableBonds(mol)
    }
df1= pd.read_csv("Canonical_smiles.output.csv")
descriptor_df = df1["Smiles"].apply(calculate_descriptors)
descriptor_df = pd.DataFrame(descriptor_df.tolist())
   
```

```python
descriptor_df
```

```python
descriptor_df.to_csv("descriptor.csv",index=False)
```

```python
descriptors_final = pd.concat([df1,descriptor_df],axis=1)
```

```python
descriptors_final
```

```python
descriptors_final.to_csv("Descriptors_final.csv",index=False)
```

```python
from rdkit import Chem
import pandas as pd
df1= pd.read_csv("Canonical_smiles.output.csv")
from rdkit.Chem import Descriptors

descriptors_values=[]
for smiles in df1["Smiles"]:
    mol=Chem.MolFromSmiles(smiles)
    if mol is not None:
        values=[descriptors[1](mol) for descriptors in Descriptors.descList]
    else:
        values= [None]*len(descriptors_names)
        
descriptors_values.append(values)
```

```python
import pandas as pd
```

```python
all_descriptors_df=pd.DataFrame(descriptors_values, columns=descriptors_names)
```
