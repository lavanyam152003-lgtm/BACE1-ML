```python
from rdkit import Chem
```

```python
import pandas as pd
```

```python
from rdkit.Chem import rdFingerprintGenerator
```

```python
from rdkit.Chem import MACCSkeys
```

```python
  from rdkit.Chem.AtomPairs import Pairs
```

```python
df1= pd.read_csv("Canonical_smiles.output.csv")
```

```python
df1
```

```python
morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2 , fpSize=2048)
```

```python
ap_gen=rdFingerprintGenerator.GetAtomPairGenerator(fpSize=2048)
```

```python
morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

morgan_fps = []
maccs_fps=[]
pubchem_fps=[]
Atom_pair_fps=[]


for smi in df1['Smiles']:
    mol = Chem.MolFromSmiles(smi)

    if mol is not None:
        morgan = morgan_gen.GetFingerprint(mol)
        morgan_fps.append(list(morgan))
        maccs = MACCSkeys.GenMACCSKeys(mol)
        maccs_fps.append(list(maccs))
        Atom_pair = ap_gen.GetFingerprint(mol)
        Atom_pair_fps.append(list( Atom_pair))
        pubchem = Chem.RDKFingerprint(mol)
        pubchem_fps.append(list(pubchem))
        
        

    else:
        morgan_fps.append([None] * 2048)
        maccs_fps.append([None] * 167)
        Atom_pair_fps.append([None] * 2048)
        pubchem_fps.append([None] * 2048)
        
```

```python
morgan_df=pd.DataFrame(morgan_fps,
                       columns=[f"morgan_fp_{i}" for i in range(2048)])
morgan_df
```

```python
MACCS_df=pd.DataFrame(maccs_fps,
                       columns=[f"maccs_fp_{i}" for i in range(167)])
MACCS_df

```

```python
pubchem_df=pd.DataFrame(pubchem_fps,
                         columns=[f"pubchem_fp_{i}" for i in range(2048)])
pubchem_df

```

```python
Atom_pair_df=pd.DataFrame(Atom_pair_fps,
                           columns=[f"atom_pair_fp_{i}" for i in range(2048)])
Atom_pair_df
```

```python
Descriptors_final = pd.read_csv("Descriptors_final.csv")
```

```python
All_fingerprints = pd.concat([df1, Descriptors_final, morgan_df, MACCS_df, pubchem_df, Atom_pair_df], axis=1)
```

```python
All_fingerprints
```

```python
All_fingerprints.to_csv("bioactivity_prediction_data.csv", index = False)
```
