```python
import pandas as pd
```

```python
df = pd.read_csv("smiles.csv")
```

```python
df.head()
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df.head()
```

```python
df.isna().sum() #to check whether there is missing values in column
```

```python
df=df.dropna(subset=['Smiles'])#remove those values
```

```python
def canonicalize_smiles(smiles): #to convert to canonical smiles
    mol = Chem.MolToSmiles(smiles)
    if mol:
        return Chem.MolToSmiles(mol,canonical=True)
    else:
         return None
    df['Canonical_SMILES'] = df['Smiles'].apply(canonical_smiles)
```

```python
df.to_csv("Canonical_smiles.output.csv", index = False)
```

```python
def is_canonical(smiles):       #just additionally to check if output is canonical or not 
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return False
    canonical = Chem.MolToSmiles(mol, canonical=True)
    return smiles == canonical
```

```python
print(is_canonical("C(CO)N"))
```

```python
df["is_canonical"] = df["Smiles"].apply(is_canonical)
```
